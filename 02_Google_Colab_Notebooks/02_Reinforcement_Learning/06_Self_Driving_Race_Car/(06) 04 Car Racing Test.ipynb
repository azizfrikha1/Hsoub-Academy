```python
from stable_baselines3 import PPO

# تصحيح اسم الملف ليطابق الملف الموجود لديك
model_file_path = "ppo_car_racing_500000.zip"

# تحميل النموذج بنجاح
model = PPO.load(model_file_path)
print("تم تحميل النموذج بنجاح!")
```

```python
# مكتبات التسجيل
from gymnasium.wrappers import RecordEpisodeStatistics, RecordVideo

# مسار مجلد الفيديويات
video_folder= './'

# عدد الحلقات
num_eval_episodes = 5

# تسجيل الفيديو
env = RecordVideo(env, video_folder=video_folder, name_prefix=f"trained_{total_timesteps}",
                  episode_trigger=lambda x: True)
# تسجيل الإحصائيات
env = RecordEpisodeStatistics(env, buffer_length=num_eval_episodes)

# الدوران في حلقات
for episode_num in range(num_eval_episodes):
    state, info = env.reset()
    # مؤشر الخروح من الحلقة
    done = False
    while not done:
      # استخدام النموذج المدرب للتنبؤ بالفعل المناسب
      action,_= model.predict(state, deterministic=True)
      # الانتقال للحالة التالية
      state, reward, terminated, truncated, info = env.step(action)
      # تحديث مؤشر الخروح من الحلقة
      if terminated or truncated:
          done = True

# طباعة الإحصائيات
print(f'Episode time taken: {env.time_queue}')
print(f'Episode total rewards: {env.return_queue}')
print(f'Episode lengths: {env.length_queue}')

# إغلاق البيئة
env.close()
```
