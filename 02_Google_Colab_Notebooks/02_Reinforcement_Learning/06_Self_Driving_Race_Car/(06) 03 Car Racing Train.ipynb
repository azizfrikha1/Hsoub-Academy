```python
import gymnasium as gym
env = gym.make("CarRacing-v3", render_mode="rgb_array")
```

```python
# [PATCHED] from google.colab import drive
# [PATCHED] drive.mount('./')
```

```python
# تضمين الصنف المطلوب لمسألتنا
from stable_baselines3 import PPO
# تضمين صنف التتبع
from stable_baselines3.common.monitor import Monitor
# تضمين صنف التعامل مع عدة بيئات
from stable_baselines3.common.vec_env import DummyVecEnv
# دالة التقييم
from stable_baselines3.common.evaluation import evaluate_policy


# استخدام صنف التتبع
env_wrapped = Monitor(env)
# استخدام صنف تعدد البيئات
env_wrapped = DummyVecEnv([lambda: env_wrapped])

# إنشاء نموذج التعلم
model = PPO("CnnPolicy", env_wrapped, verbose=1, device="cuda")

# عدد دورات التدريب
total_timesteps=500000

# تدريب النموذج
model.learn(total_timesteps=total_timesteps)

# عدد حلقات التقييم
n_eval_episodes=10

# قياس الأداء
mean_reward, std_reward = evaluate_policy(model, env_wrapped, n_eval_episodes=n_eval_episodes)
# طباعة نتائج اللتقييم
print(f"Mean reward: {mean_reward} +/- {std_reward}")
```

```python
# رفع الدرايف
# [PATCHED] from google.colab import drive
# [PATCHED] drive.mount('./')
```

```python
# مسار مجلد العمل
working_folder='./'
# اسم النموذج المدرب
model_file_name=f"ppo_car_racing_{total_timesteps}"
# مسار ملف النموذج المدرب
model_file_path = working_folder + model_file_name
model.save(model_file_path)
print("model saved .")
```
