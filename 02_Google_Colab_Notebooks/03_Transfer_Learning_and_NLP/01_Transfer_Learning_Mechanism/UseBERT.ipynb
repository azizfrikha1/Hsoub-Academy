```python
from transformers import BertModel
```

```python
from transformers import BertTokenizer
```

```python
model_name = "bert-large-uncased"
```

```python
model = BertModel.from_pretrained(model_name)
```

```python
tokenizer = BertTokenizer.from_pretrained(model_name)
```

```python
sentence1 = "I love programming"
sentence2 = "I enjoy coding"
```

```python
tokens1 = tokenizer(sentence1, return_tensors="pt")
tokens2 = tokenizer(sentence2, return_tensors="pt")
```

```python
import torch

with torch.no_grad():
    embeddings1 = model(**tokens1)["last_hidden_state"].numpy()
    embeddings2 = model(**tokens2)["last_hidden_state"].numpy()

embeddings1, embeddings2
```

```python
import numpy as np
mean_embeddings1 = np.mean(embeddings1, axis=1)
mean_embeddings2 = np.mean(embeddings2, axis=1)
mean_embeddings1,mean_embeddings2
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(mean_embeddings1, mean_embeddings2)
print("Cosine Similarity:", similarity)
```

```python
import numpy as np
import torch
from sklearn.metrics.pairwise import cosine_similarity

sentence1 = "I love programming"
sentence2 = "The movie is too bad"

# Use PyTorch tensors ("pt") instead of "tf"
tokens1 = tokenizer(sentence1, return_tensors="pt")
tokens2 = tokenizer(sentence2, return_tensors="pt")

# Pass inputs as **kwargs and convert PyTorch outputs to NumPy
with torch.no_grad():
    embeddings1 = model(**tokens1)["last_hidden_state"].numpy()
    embeddings2 = model(**tokens2)["last_hidden_state"].numpy()

# Calculate mean embeddings
mean_embeddings1 = np.mean(embeddings1, axis=1)
mean_embeddings2 = np.mean(embeddings2, axis=1)

# Compute cosine similarity
similarity = cosine_similarity(mean_embeddings1, mean_embeddings2)

print("Cosine Similarity:", similarity)
```
