```python
image_path = './people.jpg'
```

```python
img = cv2.imread(image_path)

if img is None:    raise FileNotFoundError(f"Image not found: {image_path}")
```

```python
img.shape
```

```python
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
gray_image.shape
```

```python
if not hasattr(cv2, "CascadeClassifier"):
    raise RuntimeError(
        "CascadeClassifier is not available in this OpenCV build. Install opencv-python 4.x in the notebook kernel."
    )
face_classifier = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
if face_classifier.empty():
    raise RuntimeError(
        "Failed to load Haar cascade classifier. Ensure the haarcascade XML file is present and OpenCV is installed correctly."
    )
```

```python
face = face_classifier.detectMultiScale(
    gray_image, scaleFactor=1.1, minNeighbors=5, minSize=(40, 40)
)
print("Faces detected:", len(face))
```

```python
face
```

```python
if len(face) > 0:
    for (x, y, w, h) in face:
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 4)
else:
    print("No faces were detected.")
```
