```python
import cv2
import matplotlib.pyplot as plt


def show_image(img, cmap=None):
    if img is None:
        raise ValueError("Image not loaded. Check the path.")
    if cmap:
        plt.imshow(img, cmap=cmap)
    else:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()
```

# Slicing arrays

```python
face = image[60:260, 290:420]

show_image(face)
```

# Resizing images

```python
resized = cv2.resize(image, (200, 200))

show_image(resized)
```

```python
r = 300.0 / w
dim = (300, int(h * r))
resized = cv2.resize(image, dim)
show_image(resized)
```

```python
r = 150.0 / w
dim = (150, int(h * r))
resized = cv2.resize(image, dim)
show_image(resized)
```

# Rotation images

```python
center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, -45, 1.0)
rotated = cv2.warpAffine(image, rotation_matrix, (w, h))

show_image(rotated)
```

```python
rotated = imutils.rotate(image, -45)

show_image(rotated)
```

```python
rotated = imutils.rotate_bound(image, 45)

show_image(rotated)
```

# Drawing and writing on the picture

```python
output = image.copy()
cv2.rectangle(output, (310, 90), (350, 150), (0, 0, 255), 2)

show_image(output)
```

```python
cv2.circle(output, (300, 150), 20, (255, 0, 0), -1)

show_image(output)
```

```python
output = image.copy()
cv2.line(output, (60, 20), (400, 200), (0, 0, 255), 5)

show_image(output)
```

```python
output = image.copy()
cv2.putText(output, "Hello World", (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

show_image(output)
```

# Convert image to grayscale and detect edges

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

show_image(gray, cmap='gray')
```

```python
edged = cv2.Canny(image, 30, 150)

show_image(edged, cmap='gray')
```
