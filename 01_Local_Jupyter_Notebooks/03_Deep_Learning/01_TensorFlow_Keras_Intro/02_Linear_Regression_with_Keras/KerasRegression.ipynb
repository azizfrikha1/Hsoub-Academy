```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# 1. Define Model Architecture
model = Sequential([
    Input(shape=(1,)),  # Note the tuple (1,)
    Dense(units=1, activation='linear')
])

# 2. Compile Model
model.compile(optimizer='sgd', loss='mean_absolute_error', metrics=['mae', 'mape'])

# 3. Convert Data to NumPy Arrays (Required for Keras batching)
x_train = np.array([1, 2, 3, 4], dtype=float)
y_train = np.array([2, 4, 6, 8], dtype=float)

# 4. Train
epochs = 1000
batch_size = 2

model.fit(x=x_train, y=y_train, epochs=epochs, batch_size=batch_size)
```

```python
weights, bias = model.get_weights()

print("Updated values of W and b:", round(weights[0][0],1), round(bias[0],1))
```

```python
import numpy as np
from keras.models import Sequential
from keras.layers import Input, Dense
from keras.optimizers import SGD

# 1. Define Architecture using an explicit Input layer
model = Sequential([
    Input(shape=(1,)),
    Dense(units=1, activation='linear')
])

# 2. Configure Optimizer and Compile
learning_rate = 0.02
optimizer = SGD(learning_rate=learning_rate)

model.compile(optimizer=optimizer, loss='mean_squared_error')

# 3. Convert Python lists to NumPy arrays
x_train = np.array([1, 2, 3, 4], dtype=float)
y_train = np.array([2, 4, 6, 8], dtype=float)

epochs = 1000
batch_size = 2

# 4. Train Model
model.fit(x_train, y_train, epochs=epochs, batch_size=batch_size)
```

```python
import numpy as np
import matplotlib.pyplot as plt

model = Sequential()
model.add(Dense(units=1, input_shape=(1,)))

model.compile(optimizer='sgd', loss='mean_absolute_error' , metrics=['mae','mape'])

x_train = np.array([1, 2, 3, 4])
y_train = np.array([2, 4, 6, 8])

epochs = 1000
batch_size = 2
history = model.fit(x_train, y_train, epochs=epochs, batch_size=batch_size, verbose=0)
loss_history = history.history['loss']

plt.plot(range(epochs), loss_history)
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training Loss')
plt.show()
```
